```python
# ==============================================================================
# 02_mia_evaluation.ipynb (Fashion-MNIST)
# ==============================================================================

import os
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score, roc_curve
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms


# Updated imports matching module naming
import sys
sys.path.append(os.path.join(os.getcwd(), '..'))

from src.dp_engine import run_dp_experiment

# Set plot style
plt.rcParams.update({'font.size': 12, 'figure.autolayout': True})
os.makedirs("../results/figures", exist_ok=True)
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# ------------------------------------------------------------------------------
# 1. Model Definition & Helper Functions
# ------------------------------------------------------------------------------
class FashionMLP(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(28 * 28, 128),
            nn.GroupNorm(8, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.GroupNorm(8, 64),
            nn.ReLU(),
            nn.Linear(64, num_classes)
        )
    def forward(self, x):
        return self.net(x)

def compute_sample_losses(model, loader, device):
    """Computes unreduced CrossEntropy loss for per-sample MIA evaluation."""
    model.eval()
    losses = []
    criterion = nn.CrossEntropyLoss(reduction='none')
    
    with torch.no_grad():
        for X_batch, y_batch in loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)
            logits = model(X_batch)
            loss = criterion(logits, y_batch)
            losses.extend(loss.cpu().numpy())
            
    return np.array(losses)

def evaluate_mia_attack(train_losses, test_losses):
    """Loss-based MIA: Lower loss indicates high likelihood of membership."""
    y_true = np.concatenate([np.ones_like(train_losses), np.zeros_like(test_losses)])
    scores = -np.concatenate([train_losses, test_losses]) # Higher score = lower loss
    
    mia_auc = roc_auc_score(y_true, scores)
    fpr, tpr, _ = roc_curve(y_true, scores)
    return mia_auc, fpr, tpr

# ------------------------------------------------------------------------------
# 2. Fashion-MNIST Dataset Setup
# ------------------------------------------------------------------------------
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.2860,), (0.3530,))
])

full_train_ds = datasets.FashionMNIST(root="../data", train=True, download=True, transform=transform)
test_ds = datasets.FashionMNIST(root="../data", train=False, download=True, transform=transform)


# Subsample a specified number of training samples to create a controlled memorization regime
np.random.seed(42)
training_samples = 1000


sub_indices = np.random.choice(len(full_train_ds), size=training_samples, replace=False)
train_ds_mia = Subset(full_train_ds, sub_indices)

train_loader_mia = DataLoader(train_ds_mia, batch_size=64, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=64, shuffle=False)
crit = nn.CrossEntropyLoss()

# ------------------------------------------------------------------------------
# 3. Model Training (Non-Private vs. Weak DP vs. Strong DP)
# ------------------------------------------------------------------------------

# A. Non-Private Baseline (Trained for 35 epochs to allow memorization)
print(f"--- Training Non-Private Baseline (Fashion-MNIST Subsample N={training_samples}) ---")
model_np = FashionMLP(num_classes=10).to(DEVICE)
opt_np = optim.Adam(model_np.parameters(), lr=1e-3)

model_np.train()
for epoch in range(35):
    for X_b, y_b in train_loader_mia:
        X_b, y_b = X_b.to(DEVICE), y_b.to(DEVICE)
        opt_np.zero_grad()
        loss = crit(model_np(X_b), y_b)
        loss.backward()
        opt_np.step()

tr_loss_np = compute_sample_losses(model_np, train_loader_mia, DEVICE)
te_loss_np = compute_sample_losses(model_np, test_loader, DEVICE)
mia_auc_np, fpr_np, tpr_np = evaluate_mia_attack(tr_loss_np, te_loss_np)

# B. Weak DP Model (ε ≈ 5.0 - 8.0)
print("\n--- Training Weak DP Model ---")
model_w = FashionMLP(num_classes=10).to(DEVICE)
opt_w = optim.Adam(model_w.parameters(), lr=1e-3)

_, hist_w = run_dp_experiment(
    train_loader_mia, test_loader, model_w, opt_w, crit,
    epochs=15, max_grad_norm=1.0, noise_multiplier=0.8, device=DEVICE
)
eps_w = hist_w["epsilon"][-1]

tr_loss_w = compute_sample_losses(model_w, train_loader_mia, DEVICE)
te_loss_w = compute_sample_losses(model_w, test_loader, DEVICE)
mia_auc_w, fpr_w, tpr_w = evaluate_mia_attack(tr_loss_w, te_loss_w)

# C. Strong DP Model (ε ≈ 0.5 - 1.5)
print("\n--- Training Strong DP Model ---")
model_s = FashionMLP(num_classes=10).to(DEVICE)
opt_s = optim.Adam(model_s.parameters(), lr=1e-3)

_, hist_s = run_dp_experiment(
    train_loader_mia, test_loader, model_s, opt_s, crit,
    epochs=15, max_grad_norm=1.0, noise_multiplier=2.5, device=DEVICE
)
eps_s = hist_s["epsilon"][-1]

tr_loss_s = compute_sample_losses(model_s, train_loader_mia, DEVICE)
te_loss_s = compute_sample_losses(model_s, test_loader, DEVICE)
mia_auc_s, fpr_s, tpr_s = evaluate_mia_attack(tr_loss_s, te_loss_s)

# ------------------------------------------------------------------------------
# 4. Plot MIA Vulnerability ROC Curves
# ------------------------------------------------------------------------------
plt.figure(figsize=(6.5, 5))

plt.plot(fpr_np, tpr_np, label=f'Non-Private (ε = ∞, MIA AUC = {mia_auc_np:.2f})', color='crimson', linewidth=2)
plt.plot(fpr_w, tpr_w, label=f'Weak DP (ε = {eps_w:.2f}, MIA AUC = {mia_auc_w:.2f})', color='orange', linewidth=2, linestyle='--')
plt.plot(fpr_s, tpr_s, label=f'Strong DP (ε = {eps_s:.2f}, MIA AUC = {mia_auc_s:.2f})', color='tab:blue', linewidth=2)
plt.plot([0, 1], [0, 1], 'k:', label='Random Guessing (MIA AUC = 0.50)', alpha=0.7)

plt.xlabel('False Positive Rate (Non-Member misclassified as Member)', fontweight='bold')
plt.ylabel('True Positive Rate (Member correctly identified)', fontweight='bold')
plt.title('MIA Resiliency on Fashion-MNIST', fontsize=12)
plt.legend(loc='lower right')
plt.grid(True, linestyle='--', alpha=0.5)

plt.savefig("../results/figures/mia_vulnerability_fashion.png", dpi=300, bbox_inches='tight')
plt.show()
```
